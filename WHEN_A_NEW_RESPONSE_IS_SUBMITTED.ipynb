# 1) **WHEN A NEW RESPONSE IS SUBMITTED**


👉 An action that happens automatically when someone submits (fills and sends) a form.

In simple words:
When a user clicks Submit, something else starts automatically.


**🔹 **Real-Time Example 1** *** **– Google Forms

If you create a form in Google Forms:

A student fills the form

Clicks Submit

✅ **Automatically:**

Data is saved in Google Sheets

You get an email notification

A workflow can start

This trigger is called:
“When a new response is submitted”

🔹 **Real-Time Example 2** –
**Job Application Form**

**Imagine:**

A company website has a job application form

A candidate submits resume

**Trigger:**
“When a new response is submitted”

Action:
HR gets email

Resume saved in database

Confirmation mail sent to candidate

# 2) **WHEN AN ITEM IS CREATED**

When a new row is added in a SharePoint list, the flow starts automatically.

🟢 **Example 1**: **Leave Request System**

Step 1️⃣: **You create a SharePoint List**

**List name**: **Student Leave Requests**

Title	Student Name	From Date	Reason


**Step 2️⃣**: A student adds a new entry

New row added:

| Leave 1 | Ravi | 25 Feb | Sick |

👉 This new row = one item

So now 👉 Item is created

**Step 3️⃣:** What happens in Power Automate?

Trigger:

When an item is created (SharePoint)

Action:

Send email to HOD

Send approval request

Send Teams notification

So whenever a new row is added →
Flow runs automatically.

**When an item is created Dis advantage **

Flow runs only one time:

👉 When a new row is added.

After that, if someone edits the row ❌
Flow will NOT run again.

# 3) **When item is Created or Modified**

We use it when:

👉 We want the flow to react to any change in the list.

**Example cases:**

When status changes

When manager updates comments

When price changes

When deadline changes


**🟢 Important Point**

If a row is edited:

1 time → Flow runs

2 times → Flow runs again

5 times → Flow runs 5 times

👉 Every time it is modified, the flow triggers.
